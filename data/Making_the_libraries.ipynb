```python
import os
import pandas as pd
import re
import glob
```

# Creating the libraries
Merging all the files

```python
# Find all CSV files in the folder
csv_files = glob.glob(os.path.join('DGR_profiles', '*.csv'))

# List to store DataFrames
df_list = []

for file in csv_files:
    # Extract full library code, e.g., "Lib_4A" or "Lib_4B"
    match = re.search(r'(Lib_\d+[A-Z]?)', os.path.basename(file))
    if match:
        lib = match.group(1)
        df = pd.read_csv(file)
        
        # Add Library column
        df['Library'] = lib
        
        # Append to the list
        df_list.append(df)

# Combine all DataFrames into one big DataFrame
big_df = pd.concat(df_list, ignore_index=True)
big_df.to_csv('Summary_all.csv')
```
